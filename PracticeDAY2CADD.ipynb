```python
from sklearn.preprocessing import StandardScaler

# Convert 'TotalCharges' to numeric, coercing errors (like empty strings) to NaN
x['TotalCharges'] = pd.to_numeric(x['TotalCharges'], errors='coerce')

# Fill NaN values in 'TotalCharges' with the mean of the column
x['TotalCharges'].fillna(x['TotalCharges'].mean(), inplace=True)

# Identify categorical columns (object type) for one-hot encoding
categorical_cols = x.select_dtypes(include='object').columns

# Apply one-hot encoding to the remaining categorical columns
x = pd.get_dummies(x, columns=categorical_cols, drop_first=True, dtype=int)

scaler = StandardScaler()
# Now scale the numerical features
x_scaled = scaler.fit_transform(x)
```
